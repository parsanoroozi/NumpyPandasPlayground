#### 63. Create an array class that has a name attribute (★★☆)

````python
class NamedArray(np.ndarray):
    def __new__(cls, array, name=""):
        obj = np.asarray(array).view(cls)
        obj.name = name
        return obj

    def __array_finalize__(self, obj):
        if obj is None:
            return
        self.name = getattr(obj, "name", "")

# Example
Z = NamedArray([1, 2, 3], name="MyArray")

print(Z)       # [1 2 3]
print(Z.name)  # MyArray
print(Z[1:])   # [2 3]
print(Z[1:].name)  # MyArray
````

#### 75. How to compute averages using a sliding window over an array? (★★★)

````python
def moving_average(a, n=3):
    ret = np.cumsum(a, dtype=float)
    ret[n:] = ret[n:] - ret[:-n]
    return ret[n - 1:] / n

Z = np.arange(20)
print(moving_average(Z, n=3))
# [ 1.  2.  3. ... 18.]
````

#### 78. Consider 2 sets of points P0,P1 describing lines (2d) and a point p, how to compute distance from p to each line i (P0[i],P1[i])? (★★★)

````python
def distance(P0, P1, p):
    T = P1 - P0
    L = (T**2).sum(axis=1)
    U = -((P0[:, 0] - p[..., 0]) * T[:, 0] +
          (P0[:, 1] - p[..., 1]) * T[:, 1]) / L
    U = U.reshape(len(U), 1)
    D = P0 + U * T - p
    return np.sqrt((D**2).sum(axis=1))

P0 = np.random.uniform(-10, 10, (10, 2))
P1 = np.random.uniform(-10, 10, (10, 2))
p  = np.random.uniform(-10, 10, (1, 2))
print(distance(P0, P1, p))
````

#### 79. Consider 2 sets of points P0,P1 describing lines (2d) and a set of points P, how to compute distance from each point j (P[j]) to each line i (P0[i],P1[i])? (★★★)

````python
def distances(P0, P1, P):
    T = P1 - P0                              # (n, 2) line directions
    W = P[:, None, :] - P0[None, :, :]       # (m, n, 2) vectors from each P0[i] to each P[j]
    cross = T[:, 0] * W[..., 1] - T[:, 1] * W[..., 0]   # (m, n)
    return np.abs(cross) / np.linalg.norm(T, axis=1)    # divide by line lengths, broadcasts over (n,)

P0 = np.random.uniform(-10, 10, (10, 2))
P1 = np.random.uniform(-10, 10, (10, 2))
P  = np.random.uniform(-10, 10, (5, 2))

D = distances(P0, P1, P)
print(D.shape)  # (5, 10)
````

#### 80. Consider an arbitrary array, write a function that extract a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)

````python
def extract(Z, shape, position, fill=0):
    shape = np.asarray(shape)
    position = np.asarray(position)

    start = position - shape // 2            # window start in Z's coordinates (may be negative)
    stop = start + shape

    R = np.full(shape, fill, dtype=Z.dtype)

    # Overlap between the window and Z, in Z's coordinates
    z_start = np.maximum(start, 0)
    z_stop = np.minimum(stop, Z.shape)
    if np.any(z_stop <= z_start):            # window lies entirely outside Z
        return R

    # The same overlap, in R's coordinates
    r_start = z_start - start
    r_stop = r_start + (z_stop - z_start)

    z_slices = tuple(slice(a, b) for a, b in zip(z_start, z_stop))
    r_slices = tuple(slice(a, b) for a, b in zip(r_start, r_stop))
    R[r_slices] = Z[z_slices]
    return R

Z = np.arange(25).reshape(5, 5)
print(extract(Z, shape=(3, 3), position=(0, 0)))
# [[0 0 0]
#  [0 0 1]
#  [0 5 6]]

print(extract(Z, shape=(3, 3), position=(2, 2)))
# [[ 6  7  8]
#  [11 12 13]
#  [16 17 18]]
````

````python
class Symmetric(np.ndarray):
    def __setitem__(self, index, value):
        i, j = index
        super().__setitem__((i, j), value)
        super().__setitem__((j, i), value)

def symmetric(Z):
    return np.asarray(Z + Z.T - np.diag(Z.diagonal())).view(Symmetric)

S = symmetric(np.random.randint(0, 10, (5, 5)))
S[2, 3] = 42
print(S[2, 3], S[3, 2])   # 42 42
print(np.array_equal(S, S.T))  # True
````

#### 88. How to implement the Game of Life using numpy arrays? (★★★)

````python

def iterate(Z):
    # Count neighbours of each interior cell by summing 8 shifted slices
    N = (Z[0:-2, 0:-2] + Z[0:-2, 1:-1] + Z[0:-2, 2:] +
         Z[1:-1, 0:-2]                 + Z[1:-1, 2:] +
         Z[2:  , 0:-2] + Z[2:  , 1:-1] + Z[2:  , 2:])

    # Apply the rules
    birth = (N == 3) & (Z[1:-1, 1:-1] == 0)
    survive = ((N == 2) | (N == 3)) & (Z[1:-1, 1:-1] == 1)

    Z[...] = 0
    Z[1:-1, 1:-1][birth | survive] = 1
    return Z

Z = np.random.randint(0, 2, (50, 50))
for _ in range(100):
    Z = iterate(Z)
````

#### 90. Given an arbitrary number of vectors, build the cartesian product (every combinations of every item) (★★★)

````python
def cartesian(*arrays):
    grids = np.meshgrid(*arrays, indexing="ij")
    return np.stack(grids, axis=-1).reshape(-1, len(arrays))

print(cartesian([1, 2, 3], [4, 5], [6, 7]))
# [[1 4 6]
#  [1 4 7]
#  [1 5 6]
#  [1 5 7]
#  [2 4 6]
#  ...
#  [3 5 7]]
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?

````python
# Example path: an Archimedean spiral
phi = np.arange(0, 10 * np.pi, 0.1)
x = phi * np.cos(phi)
y = phi * np.sin(phi)

# 1. Length of each segment between consecutive points
dr = np.sqrt(np.diff(x)**2 + np.diff(y)**2)

# 2. Cumulative distance along the path at each point
r = np.zeros_like(x)
r[1:] = np.cumsum(dr)

# 3. Evenly spaced distances along the path
r_int = np.linspace(0, r[-1], 200)

# 4. Interpolate x and y at those distances
x_int = np.interp(r_int, r, x)
y_int = np.interp(r_int, r, y)
````
