```python
climate_data = np.genfromtxt('files/climate.txt', delimiter=',', skip_header=1)
```

```python
climate_data
```

```python
climate_data.shape
```

```python
weights = np.array([0.3, 0.2, 0.5])
```

```python
yields = climate_data @ weights
```

```python
yields
```

```python
yields.shape
```

```python
climate_result = np.concatenate((climate_data, yields.reshape(10000,1)), axis=1)
```
